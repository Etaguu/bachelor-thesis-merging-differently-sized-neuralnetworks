```python
from model.Create_Neural_Network import *
from utils.DataLoader import get_task_loaders,load_dataset
from merging.Fisher_Information_approximation_Library import *
from merging.Fisher_Neuron_Select_Strategy import *
from utils.Seeding import set_seed
from utils.Results_File import save_results
from tabels.Create_Figures import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
train_complete_raw_mnist, test_complete_raw_mnist = load_dataset(dataset_name="MNIST")
train_complete_loader_mnist, test_complete_loader_mnist = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=True)
```

```python
specialist_results = {
        "Neural Network (8 Neurons)": [],
        "Neural Network (64 Neurons)": [],
        "Neural Network (256 Neurons)": []
}

pretraining_results = {
        "Neural Network (8 Neurons)": [],
        "Neural Network (64 Neurons)": [],
        "Neural Network (256 Neurons)": []
}

seeds = [0,1,2,3,4]

for seed in seeds:
    set_seed(seed)

    neural_network_8 = NeuralNetwork8().to(DEVICE)
    neural_network_64 = NeuralNetwork64().to(DEVICE)
    neural_network_256 = NeuralNetwork256().to(DEVICE)    

    accuracy_8_pre = evaluate(neural_network_8, test_complete_loader_mnist)
    accuracy_64_pre = evaluate(neural_network_64, test_complete_loader_mnist)
    accuracy_256_pre = evaluate(neural_network_256, test_complete_loader_mnist)
    pretraining_results["Neural Network (8 Neurons)"].append(accuracy_8_pre)    
    pretraining_results["Neural Network (64 Neurons)"].append(accuracy_64_pre)   
    pretraining_results["Neural Network (256 Neurons)"].append(accuracy_256_pre)


    train(neural_network_8,train_complete_loader_mnist,10,"Small Model")
    accuracy_8 = evaluate(neural_network_8, test_complete_loader_mnist)
    specialist_results["Neural Network (8 Neurons)"].append(accuracy_8)    


    train(neural_network_64,train_complete_loader_mnist,10, "Middle Model")
    accuracy_64 = evaluate(neural_network_64, test_complete_loader_mnist)
    specialist_results["Neural Network (64 Neurons)"].append(accuracy_64)   

    train(neural_network_256,train_complete_loader_mnist,10, "Big Model")
    accuracy_256 = evaluate(neural_network_256, test_complete_loader_mnist)
    specialist_results["Neural Network (256 Neurons)"].append(accuracy_256)


# Save results AND seeds
data = {
    "specialist results": specialist_results,
    "pre results": pretraining_results
}

with open("baseline_mnist_all_sizes.pkl", "wb") as f:
    pickle.dump(data, f)

print("Results saved!")
```
