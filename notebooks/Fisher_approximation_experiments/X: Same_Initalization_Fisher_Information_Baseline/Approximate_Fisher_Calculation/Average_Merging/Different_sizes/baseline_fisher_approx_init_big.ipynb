```python
from tabels.Create_Figures import create_accuracy_table
from model.Create_Neural_Network import *
from utils.DataLoader import get_task_loaders,load_dataset
from model.Define_Experiment_Fisher import *
from merging.Fisher_Information_approximation_Library import *
from merging.Fisher_Neuron_Select_Strategy import *
import numpy as np
import matplotlib.pyplot as plt
```

```python
train_complete_raw_mnist, test_complete_raw_mnist = load_dataset(dataset_name="MNIST")
train_complete_loader_mnist, test_complete_loader_mnist = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=True)
train_complete_raw_fashion, test_complete_raw_fashion = load_dataset(dataset_name="FASHION_MNIST")
train_complete_loader_fashion, test_complete_loader_fashion = get_task_loaders(batch_size=64,dataset_name="FASHION_MNIST",do_shuffle=True)
```

```python

results = {
    "Merged Model (No Mask)": {"mnist": [], "fashion": []},  
    "Merged Model (Mask)":    {"mnist": [], "fashion": []},  
}

specialist_results = {
    "big_model": {
        "fashion": [],
        "mnist": []
    },
    "small_model": {
        "fashion": [],
        "mnist": []
    },
    "small_random": {
        "Fashion (Pretraining)": [],
        "MNIST: (random init)": []
    }
}

run_experiment_fisher(
    model_classes=[NeuralNetwork512, NeuralNetwork256],
    train_raw=[train_complete_raw_fashion,train_complete_raw_mnist],
    train_loaders=[train_complete_loader_fashion,train_complete_loader_mnist],
    test_loaders=[test_complete_loader_fashion,test_complete_loader_mnist],
    seeds=[0, 1, 2, 3, 4],
    specialist_results=specialist_results,
    results=results,
    epochs=10,
    learn_rate=1e-3,
    fisher_n=1000,
    neuron_select_strategy="random",
    merge_method="average",
    save_path="results/baseline_mnist_nn512_average_random.pkl"
)
```

```python
create_accuracy_table(specialist_results=specialist_results,results=results)

plt.tight_layout()
plt.show()
```
