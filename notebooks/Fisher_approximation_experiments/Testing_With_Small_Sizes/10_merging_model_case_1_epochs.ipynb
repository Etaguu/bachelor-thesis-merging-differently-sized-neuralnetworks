```python
from tabels.Create_Figures import create_accuracy_table
from model.Create_Neural_Network import *
from utils.DataLoader import get_task_loaders,load_dataset
from model.Define_Experiment import *
import matplotlib.pyplot as plt
```

```python
train_complete_raw, test_complete_raw = load_dataset()
train_complete_loader, test_complete_loader = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=False)
```

```python
# Not shuffled, increasing the epochs, increase the accuracy, simple as that

results_5 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_10 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_20 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_40 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

specialist_results_5 = {
        "Model A": [],
        "Model B": [],
}

specialist_results_10 = {
        "Model A": [],
        "Model B": [],
}

specialist_results_20 = {
        "Model A": [],
        "Model B": [],
}
specialist_results_40 = {
        "Model A": [],
        "Model B": [],
}
```

```python
print("Tasks: Training iteratively — Same initialization, same data, same parameter count, not shuffled")


run_experiment(
    model_factory=create_same_size_same_init,
    model_classes=[NeuralNetwork8, NeuralNetwork8],
    train_raw=[train_complete_raw, train_complete_raw],
    train_loaders=[train_complete_loader, train_complete_loader],
    test_loaders=test_complete_loader,
    seeds=[1,2,3,4,5],
    specialist_results=specialist_results_5,
    merge_results=results_5,
    epochs=5,
    fisher_n=1000,
    learn_rate=1e-3,
    save_path="results/10_mnist_epochs_5.pkl"
)
```

```python
create_accuracy_table(specialist_results_5,results_5)

plt.title(
    "Model and Merge Performance (5 seeds)\n"
    "same initialization, same data, same parameter count\n"
    "MNIST, not shuffled, Very Tiny Neural Network",
    fontweight="bold",
    pad=15,
)
plt.tight_layout()
plt.show()
```
