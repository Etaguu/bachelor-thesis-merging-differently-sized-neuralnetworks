```python
from tabels.Create_Figures import create_accuracy_table
from model.Create_Neural_Network import *
from utils.DataLoader import get_task_loaders,load_dataset
from model.Define_Experiment import *
import matplotlib.pyplot as plt
```

```python
train_complete_raw, test_complete_raw = load_dataset()
train_complete_loader, test_complete_loader = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=False)
```

```python
#Test the difference between different sizes of NN and shuffeling the data, in the tiny networks it has a real
# disadvantage and makes the performance of the specialist and merged model way worse, not happening in big NN
results_very_tiny = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_tiny = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_small = {
        "Average Merging": [],
        "Fisher Merging": [],
}

specialist_results_very_tiny = {
        "Model A": [],
        "Model B": [],
}

specialist_results_tiny = {
        "Model A": [],
        "Model B": [],
}
specialist_results_small = {
        "Model A": [],
        "Model B": [],
}
```

```python
print("Tasks: Training iteratively — Same initialization, same data, same parameter count, not shuffled")

run_experiment(
    model_factory=create_same_size_same_init,
    model_classes=[NeuralNetwork8, NeuralNetwork8],
    train_raw=[train_complete_raw, train_complete_raw],
    train_loaders=[train_complete_loader, train_complete_loader],
    test_loaders=test_complete_loader,
    seeds=[1,2,3,4,5],
    specialist_results=specialist_results_very_tiny,
    merge_results=results_very_tiny,
    epochs=10,
    fisher_n=1000,
    save_path="results/13_mnist_no_shuffle_very_tiny.pkl"
)
```

```python
run_experiment(
    model_factory=create_same_size_same_init,
    model_classes=[NeuralNetwork16, NeuralNetwork16],
    train_raw=[train_complete_raw, train_complete_raw],
    train_loaders=[train_complete_loader, train_complete_loader],
    test_loaders=test_complete_loader,
    seeds=[1,2,3,4,5],
    specialist_results=specialist_results_tiny,
    merge_results=results_tiny,
    epochs=10,
    fisher_n=1000,
    save_path="results/13_no_mnist_shuffle_16.pkl"
)
```

```python
run_experiment(
    model_factory=create_same_size_same_init,
    model_classes=[NeuralNetwork256, NeuralNetwork256],
    train_raw=[train_complete_raw, train_complete_raw],
    train_loaders=[train_complete_loader, train_complete_loader],
    test_loaders=test_complete_loader,
    seeds=[1,2,3,4,5],
    specialist_results=specialist_results_small,
    merge_results=results_small,
    epochs=10,
    fisher_n=1000,
    save_path="results/13_mnist_no_shuffle_256.pkl"
)
```
