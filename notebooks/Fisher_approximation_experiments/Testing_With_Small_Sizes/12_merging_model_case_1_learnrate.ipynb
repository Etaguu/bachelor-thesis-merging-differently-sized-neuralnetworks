```python
from tabels.Create_Figures import create_accuracy_table
from model.Create_Neural_Network import *
from utils.DataLoader import get_task_loaders,load_dataset
from model.Define_Experiment import *
import matplotlib.pyplot as plt
```

```python
train_complete_raw, test_complete_raw = load_dataset()
train_complete_loader, test_complete_loader = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=False)
```

```python
# The best learn rate is 0.001 as already used in the other experiments, a larger or smaller one lead to less accurate results

results_01 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_02 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_03 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_04 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

specialist_results_01 = {
        "Model A": [],
        "Model B": [],
}

specialist_results_02 = {
        "Model A": [],
        "Model B": [],
}

specialist_results_03 = {
        "Model A": [],
        "Model B": [],
}
specialist_results_04 = {
        "Model A": [],
        "Model B": [],
}
```

```python
run_experiment(
    model_factory=create_same_size_same_init,
    model_classes=[NeuralNetwork8, NeuralNetwork8],
    train_raw=[train_complete_raw, train_complete_raw],
    train_loaders=[train_complete_loader, train_complete_loader],
    test_loaders=test_complete_loader,
    seeds=[1,2,3,4,5],
    specialist_results=specialist_results_01,
    merge_results=results_01,
    epochs=10,
    fisher_n=1000,
    learn_rate=1e-2,
    save_path="results/12_mnist_learnrate_small.pkl"
)
```

```python
create_accuracy_table(specialist_results_01,results_01)

plt.title(
    "Model and Merge Performance (5 seeds)\n"
    "same initialization, same data, same parameter count\n"
    "MNIST, not shuffled, Very Tiny Neural Network Leanrate = 0.01",
    fontweight="bold",
    pad=15,
)
plt.tight_layout()
plt.show()
```
