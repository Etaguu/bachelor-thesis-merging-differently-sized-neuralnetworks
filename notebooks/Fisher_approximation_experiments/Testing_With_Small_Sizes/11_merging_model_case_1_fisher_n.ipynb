```python
from tabels.Create_Figures import create_accuracy_table
from model.Create_Neural_Network import *
from utils.DataLoader import get_task_loaders,load_dataset
from model.Define_Experiment import *
import matplotlib.pyplot as plt
```

```python
train_complete_raw, test_complete_raw = load_dataset()
train_complete_loader, test_complete_loader = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=False)
```

```python
#Calculates on the TinyNetwork how important the fisher batch parameter is, for MNIST it is not important
# in the usage of the approximate fisher merging

results_10 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_100 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_1000 = {
        "Average Merging": [],
        "Fisher Merging": [],
}

results_whole = {
        "Average Merging": [],
        "Fisher Merging": [],
}

specialist_results_10 = {
        "Model A": [],
        "Model B": [],
}

specialist_results_100 = {
        "Model A": [],
        "Model B": [],
}

specialist_results_1000 = {
        "Model A": [],
        "Model B": [],
}
specialist_results_whole = {
        "Model A": [],
        "Model B": [],
}
```

```python
print("Tasks: Training iteratively — Same initialization, same data, same parameter count, not shuffled")

run_experiment(
    model_factory=create_same_size_same_init,
    model_classes=[NeuralNetwork16, NeuralNetwork16],
    train_raw=[train_complete_raw, train_complete_raw],
    train_loaders=[train_complete_loader, train_complete_loader],
    test_loaders=test_complete_loader,
    seeds=[1,2,3,4,5],
    specialist_results=specialist_results_10,
    merge_results=results_10,
    epochs=10,
    fisher_n=10,
    save_path="results/11_mnist_fisher_n_10.pkl"
)
```
