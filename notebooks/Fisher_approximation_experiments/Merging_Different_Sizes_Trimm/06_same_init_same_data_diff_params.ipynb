```python
train_complete_raw_mnist, test_complete_raw_mnist = load_dataset(dataset_name="MNIST")
train_complete_loader_mnist, test_complete_loader_mnist = get_task_loaders(batch_size=64,dataset_name="MNIST",do_shuffle=True)
train_complete_raw_fashion, test_complete_raw_fashion = load_dataset(dataset_name="FASHION_MNIST")
train_complete_loader_fashion, test_complete_loader_fashion = get_task_loaders(batch_size=64,dataset_name="FASHION_MNIST",do_shuffle=True)
```
