```python
from tabels.Save_Results import *
from tabels.Create_Figures import *
```

```python
columns_order = ["Model Big (Full)", "Model Big (Trimmed)", "Model Small", "Average Merging", "Fisher Merging"]
columns       = ["Experiment", "Model Big (Full)", "Model Big (Trimmed)", "Model Small", "Average Merging", "Fisher Merging"]
```

```python
basic_experiments = [
    "06_fashion_bad_fisher.pkl",
    "05_fashion_bad_fisher.pkl",
]

experiment_names = [
    "Same initialization",
    "Different initialization",
]

figure = create_comparison_table_from_files(
    files=basic_experiments,
    experiment_names=experiment_names,
    column_names=columns,
    column_order=columns_order
)

figure.show()

fig = plot_bar_comparison(
    files=basic_experiments,
    experiment_names=experiment_names,
    column_names=columns,
    column_order=columns_order
)
plt.show()
```
